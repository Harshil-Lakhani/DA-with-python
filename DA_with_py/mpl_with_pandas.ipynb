# Matplotlib with pandas

```python
import pandas as pd
import matplotlib.pyplot as plt
data={
    'Month':['JAN','FEB','MAR','APRIL'],
    'Sales':[25000,20000,25000,35000]
}
df=pd.DataFrame(data)

plt.bar(df['Month'],df['Sales'])
plt.xlabel('Months')
plt.ylabel('Sales')
plt.title('Monthly Report of sales department')
plt.show()
```
