# subplot

```python
import matplotlib.pyplot as plt
cat=['a','b','c','d']
val=[10,20,40,30]

x=[1,2,3,4,5]
y=[9,8,7,6,5]

plt.subplot(2,2,1)
plt.bar(x,y)

plt.subplot(2,2,2)
plt.scatter(cat,val)

plt.subplot(2,2,3)
plt.scatter(cat,val)

plt.subplot(2,2,4)
plt.bar(x,y)
plt.show
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data={
    'mon':['jan','feb','march','april'],
    'sales':[12000,11000,13000,25000]
}
df=pd.DataFrame(data)
plt.bar(df['mon'],df['sales'])
plt.title('Matplotlib with pandas')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y1=[20,23,35,65,70]
y2=[30,40,50,70,80]

plt.plot(x,y1,label='First')
plt.plot(x,y2,label='Second')
plt.legend()
plt.grid()
plt.show()
```
