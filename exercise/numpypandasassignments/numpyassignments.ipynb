### Assignment 1: Array Creation and Manipulation

1. Create a NumPy array of shape (5, 5) filled with random integers between 1 and 20. Replace all the elements in the third column with 1.


```python
import numpy as np
arr1=np.random.randint(1,21,size=(5,5))
print(f"arr1={arr1}")
arr1[:,2]=1
print("\n")
print(f"new arr1={arr1}")
```

2. Create a NumPy array of shape (4, 4) with values from 1 to 16. Replace the diagonal elements with 0.


```python
import numpy as np
arr2=np.arange(1,17).reshape((4,4))
np.fill_diagonal(arr2,0)
print(arr2)
```


### Assignment 2: Array Indexing and Slicing

1. Create a NumPy array of shape (6, 6) with values from 1 to 36. Extract the sub-array consisting of the 3rd to 5th rows and 2nd to 4th columns.


```python
import numpy as np
arr1=np.arange(1,37).reshape((6,6))
sub=arr1[2:5,1:4]
print(sub)
```

2. Create a NumPy array of shape (5, 5) with random integers. Extract the elements on the border.`

```python
# Create a NumPy array of shape (5, 5) with random integers
array = np.arange(1,26).reshape(5,5)
print("Original array:")
print(array)



# Extract the elements on the border
border_elements = np.concatenate((array[0, :], array[-1, :], array[1:-1, 0], array[1:-1, -1]))
print("Border elements:")
print(border_elements)
```

### Assignment 3: Array Operations

1. Create two NumPy arrays of shape (3, 4) filled with random integers. Perform element-wise addition, subtraction, multiplication, and division.


```python
arr1=np.random.randint(1,21,size=(3,4))
arr2=np.random.randint(1,21,size=(3,4))
print(arr1)
print(arr2)
print(f"addition=\n{arr1+arr2}")
print(f"sub=\n{arr1-arr2}")
print(f"mul=\n{arr1*arr2}")
print(f"div=\n{arr1/arr2}")
```

2. Create a NumPy array of shape (4, 4) with values from 1 to 16. Compute the row-wise and column-wise sum.

```python
arr=np.arange(1,17).reshape(4,4)
print(arr)
row_sum=np.sum(arr,axis=1)
col_sum=np.sum(arr,axis=0)
print(f"sum of row={row_sum}")
print(f"sum of col={col_sum}")
```
